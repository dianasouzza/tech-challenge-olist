# Explorando a Recompra (~3%)

## Cada pedido tem um customer_id novo, mesmo que a pessoa seja a mesma. A pessoa de verdade é o customer_unique_id. A taxa de recompra é: (pessoas com mais de 1 pedido) ÷ (total de pessoas).

```python
#Passo 1: entender as duas chaves

print(len(df_customers))                           # esperado: 99441 (uma linha por pedido)
print(df_customers['customer_id'].nunique())       # esperado: 99441
print(df_customers['customer_unique_id'].nunique())  # esperado: cerca de 96 mil
```

Customer_unique_id é menor que customer_id, é porque algumas pessoas aparecem em mais de um pedido. Essa diferença é a recompra. unique_id tem apenas os clientes de forma distinta. customer_id é associado ao pedido.

```python
#Passo 2: levar o customer_unique_id para os pedidos
df_cli = df_orders.merge(df_customers[['customer_id','customer_unique_id']], on='customer_id', how='left')
print(len(df_cli))   # esperado: 99441
```

```python
#Passo 3: quantos pedidos cada pessoa fez

pedidos_por_cliente = df_cli.groupby('customer_unique_id').size()
pedidos_por_cliente.value_counts().sort_index()
```

```python
#Passo 4: a taxa de recompra

total = len(pedidos_por_cliente)
recompra = (pedidos_por_cliente > 1).sum()
print(total, recompra, round(recompra / total * 100, 1))
```

```python
# Passo 5: testar com só os pedidos entregues

entregues = df_cli[df_cli['order_status'] == 'delivered']
ped_ent = entregues.groupby('customer_unique_id').size()
print(len(ped_ent), (ped_ent > 1).sum(), round((ped_ent > 1).sum() / len(ped_ent) * 100, 1))
```

96.096 é o total de clientes únicos (pessoas) 2.997 são pessoas que fizeram mais de 1 pedido, e 3,1% é 2.997 ÷ 96.096.<br>

 3,1% dos clientes compraram mais de uma vez
